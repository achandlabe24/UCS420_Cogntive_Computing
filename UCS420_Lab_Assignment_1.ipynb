```python
#Assingment 5.1: WAP to find max amoung three numbers and input from user. [Using max() function]
p = int(input("Enter first number: "))
q = int(input("Enter second number: "))
r = int(input("Enter third number: "))

maximum = max(p, q, r)
print(f"{maximum} is the largest number.")

#Assingment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.
n = int(input("\nEnter number for divisible check: "))
div_sum = 0
for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        print(f"{i} is divisible by 7 and 9")
        div_sum += i
print(f"Sum of these numbers: {div_sum}")

#Assingment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.
n_prime = int(input("\nEnter number to find primes up to: "))
prime_sum = 0
for num in range(2, n_prime + 1):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum += num

print(f"The sum of all prime numbers up to {n_prime} is {prime_sum}")
```

```python
#Assingment 6.1: WAP using function that add all odd numbers from 1 to n, n is given by the user.
def sum_odd_numbers(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total

n1 = int(input("Enter n to sum odd numbers: "))
print(f"Sum of odd numbers up to {n1} is: {sum_odd_numbers(n1)}")

#Assingment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.
def is_prime(num):
    if num < 2: return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def sum_prime_numbers(n):
    total = 0
    for i in range(2, n + 1):
        if is_prime(i):
            total += i
    return total

n2 = int(input("\nEnter n to sum prime numbers: "))
print(f"Sum of prime numbers up to {n2} is: {sum_prime_numbers(n2)}")
```
