## Question 1: Plotting Mathematical Functions with Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

# Ask the user to enter a value for M
M = float(input("Enter a value for M: "))

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 400)

# Compute y values
y1 = M * (x**2)  # y = M * x^2
y2 = M * np.sin(x)  # y = M * sin(x)

# Create plot
plt.figure(figsize=(9, 5))

# Plot both functions with different colors and line styles
plt.plot(x, y1, color="crimson", linestyle="--", label=f"$y = {M}x^2$")
plt.plot(
    x, y2, color="royalblue", linestyle="-", label=f"$y = {M}\\sin(x)$"
)

# Customizations
plt.title(f"Mathematical Functions for M = {M}", fontsize=14, fontweight="bold")
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.7)
plt.legend(loc="best")

# Show the plot
plt.show()
```

## Question 2: Seaborn Bar Plot with Subject Scores

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Create dataset of five subjects and scores
data = {
    "Subject": [
        "Mathematics",
        "Data Structures",
        "Operating Systems",
        "Python",
        "Machine Learning",
    ],
    "Score": [88, 92, 79, 95, 85],
}

# Convert into a Pandas DataFrame
df = pd.DataFrame(data)

# Create Seaborn bar plot
plt.figure(figsize=(8, 5))
ax = sns.barplot(
    x="Subject", y="Score", data=df, hue="Subject", palette="Set2", legend=False
)

# Annotations on top of each bar
for bar in ax.patches:
    height = bar.get_height()
    ax.annotate(
        f"{int(height)}",
        xy=(bar.get_x() + bar.get_width() / 2, height),
        xytext=(0, 3),  # 3 points vertical offset
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold",
    )

# Title, axis labels, and grid
plt.title("Subject Scores Overview", fontsize=14, fontweight="bold")
plt.xlabel("Subjects", fontsize=12)
plt.ylabel("Scores", fontsize=12)
plt.ylim(0, 105)  # Leave room for annotations
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

## Question 3: $2 \times 2$ Subplot Layout with Seed and Random Noise
Note: Set roll_number to your actual roll number.

```python
import matplotlib.pyplot as plt
import numpy as np

# Set roll number as seed for NumPy reproducibility
roll_number = 1024170259
np.random.seed(roll_number)

# Generate a dataset of 50 values using np.random.randn()
data = np.random.randn(50)

# Create a 2x2 subplot layout
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Subplot 1: Line plot showing cumulative sum
cum_sum = np.cumsum(data)
axes[0, 0].plot(cum_sum, color="teal", marker="o", markersize=4)
axes[0, 0].set_title("Cumulative Sum Line Plot", fontweight="bold")
axes[0, 0].set_xlabel("Index")
axes[0, 0].set_ylabel("Cumulative Sum")
axes[0, 0].grid(True, linestyle="--", alpha=0.6)

# Subplot 2: Scatter plot with random noise
noise = np.random.normal(0, 0.5, 50)
scatter_data = data + noise
axes[0, 1].scatter(
    range(50), scatter_data, color="darkorange", alpha=0.8, edgecolors="k"
)
axes[0, 1].set_title("Scatter Plot with Random Noise", fontweight="bold")
axes[0, 1].set_xlabel("Index")
axes[0, 1].set_ylabel("Value + Noise")
axes[0, 1].grid(True, linestyle="--", alpha=0.6)

# Subplot 3: Original Data Line Plot
axes[1, 0].plot(data, color="purple", linestyle="-.")
axes[1, 0].set_title("Original Random Data", fontweight="bold")
axes[1, 0].set_xlabel("Index")
axes[1, 0].set_ylabel("Value")
axes[1, 0].grid(True, linestyle="--", alpha=0.6)

# Subplot 4: Histogram of Values
axes[1, 1].hist(data, bins=10, color="forestgreen", edgecolor="black", alpha=0.7)
axes[1, 1].set_title("Data Distribution (Histogram)", fontweight="bold")
axes[1, 1].set_xlabel("Value")
axes[1, 1].set_ylabel("Frequency")
axes[1, 1].grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

## Question 4: Seaborn Visualization on Company Sales Dataset

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# GitHub raw URL for the dataset
url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
df = pd.read_csv(url)

# Set Seaborn theme style
sns.set_theme(style="whitegrid")

# -------------------------------------------------------------
# 1. Total profit of all months using a line plot
# -------------------------------------------------------------
plt.figure(figsize=(9, 4))
sns.lineplot(
    data=df,
    x="month_number",
    y="total_profit",
    marker="o",
    color="green",
    linewidth=2.5,
)
plt.title(
    "Total Profit of All Months", fontsize=14, fontweight="bold", pad=12
)
plt.xlabel("Month Number", fontsize=11)
plt.ylabel("Total Profit ($)", fontsize=11)
plt.xticks(df["month_number"])
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
# 2. All product sales data using a multiline plot
# -------------------------------------------------------------
# Identify product columns (excluding non-product aggregate columns)
product_cols = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer",
]

plt.figure(figsize=(10, 5))
for col in product_cols:
    sns.lineplot(
        data=df, x="month_number", y=col, label=col.capitalize(), marker="o"
    )

plt.title(
    "All Product Sales Data Monthly Comparison",
    fontsize=14,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Month Number", fontsize=11)
plt.ylabel("Units Sold", fontsize=11)
plt.xticks(df["month_number"])
plt.legend(title="Products", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
# 3. Bar chart for all features/attributes
# -------------------------------------------------------------
# Reshape dataframe to long format for easy plotting in Seaborn
df_melted = df.melt(
    id_vars=["month_number"], var_name="Attribute", value_name="Value"
)

plt.figure(figsize=(12, 6))
sns.barplot(data=df_melted, x="Attribute", y="Value", palette="viridis")
plt.title(
    "Bar Chart for All Dataset Attributes",
    fontsize=14,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Attributes / Features", fontsize=11)
plt.ylabel("Values", fontsize=11)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
