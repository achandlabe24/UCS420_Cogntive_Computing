```python
import pandas as pd

# ==============================================================================
# Q1: Create DataFrame
# ==============================================================================
data_q1 = {
    'Refund': ['Yes', 'No', 'No', 'Yes', 'No', 'No', 'Yes', 'No', 'No', 'No'],
    'Marital Status': ['Single', 'Married', 'Single', 'Married', 'Divorced', 'Married', 'Divorced', 'Single', 'Married', 'Single'],
    'Taxable Income': ['125K', '100K', '70K', '120K', '95K', '60K', '220K', '85K', '75K', '90K'],
    'Cheat': ['No', 'No', 'No', 'No', 'Yes', 'No', 'No', 'Yes', 'No', 'Yes']
}

df1 = pd.DataFrame(data_q1)
print("=== Q1: Created DataFrame ===")
print(df1)
print("\n" + "="*60 + "\n")
```

```python
# ==============================================================================
# Q3: Navigate DataFrame
# ==============================================================================
print("=== Q3.1: Select rows from index 3 to 7 ===")
print(df1.iloc[3:8])

print("\n=== Q3.2: Select rows from index 4 to 8, and columns 2 to 4 ===")
# iloc index slices: rows 4:9 (4 to 8), cols 2:5 (index 2 to 4)
print(df1.iloc[4:9, 2:5])

print("\n=== Q3.3: Select all rows with column index 1 to 3 (including index 3) ===")
print(df1.iloc[:, 1:4])
print("\n" + "="*60 + "\n")
```

```python

# ==============================================================================
# Q5: Delete row 4 and column index 3
# ==============================================================================
# Delete row 4 using drop(index=4) and column index 3 using df.columns[3]
df_iris_modified = df_iris.drop(index=4).drop(columns=df_iris.columns[3])

print("=== Q5: Iris dataset after deleting Row 4 & Column Index 3 ===")
print(df_iris_modified.head())
print("\n" + "="*60 + "\n")
```

```python
# ==============================================================================
# Q6: Employees Dataset Analysis
# ==============================================================================
emp_data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 85000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-05-19', '2015-08-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 8000, 4500, 6000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv('employees.csv', index=False)

# (a) Shape of DataFrame
print("=== Q6(a): Shape (Rows, Columns) ===")
print(emp_df.shape)

# (b) Summary of DataFrame
print("\n=== Q6(b): DataFrame Info & Summary ===")
emp_df.info()

# (c) Descriptive statistics
print("\n=== Q6(c): Descriptive Statistics ===")
print(emp_df.describe())

# (d) First 5 rows and last 3 rows
print("\n=== Q6(d): First 5 Rows ===")
print(emp_df.head(5))
print("\n=== Q6(d): Last 3 Rows ===")
print(emp_df.tail(3))

# (e) Calculate statistics
print("\n=== Q6(e): Specific Statistics ===")
print(f"i.   Average Salary: {emp_df['Salary'].mean()}")
print(f"ii.  Total Bonus Paid: {emp_df['Bonus'].sum()}")
print(f"iii. Youngest Employee Age: {emp_df['Age'].min()}")
print(f"iv.  Highest Performance Rating: {emp_df['Rating'].max()}")

# (f) Sort by Salary in descending order
print("\n=== Q6(f): Sorted by Salary (Descending) ===")
print(emp_df.sort_values(by='Salary', ascending=False))

# (g) Categorize employees based on rating
def get_performance_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

emp_df['Performance_Category'] = emp_df['Rating'].apply(get_performance_category)
print("\n=== Q6(g): Performance Categorization ===")
print(emp_df[['Name', 'Rating', 'Performance_Category']])

# (h) Identify missing values
print("\n=== Q6(h): Missing Values Count per Column ===")
print(emp_df.isnull().sum())

# (i) Rename Employee_ID column to ID
emp_df.rename(columns={'Employee_ID': 'ID'}, inplace=True)
print("\n=== Q6(i): Renamed Column ('Employee_ID' -> 'ID') ===")
print(emp_df.columns.tolist())

# (j) Filtering conditions
print("\n=== Q6(j.i): Employees with > 5 Years of Experience ===")
print(emp_df[emp_df['Years_of_Experience'] > 5])

print("\n=== Q6(j.ii): Employees in IT Department ===")
print(emp_df[emp_df['Department'] == 'IT'])

# (k) Add Tax column (10% deduction from Salary)
emp_df['Tax'] = emp_df['Salary'] * 0.10
print("\n=== Q6(k): Dataset with Added Tax Column ===")
print(emp_df[['Name', 'Salary', 'Tax']])

# (l) Save modified DataFrame to a new CSV file
emp_df.to_csv('modified_employees.csv', index=False)
print("\n=== Q6(l): Modified DataFrame saved to 'modified_employees.csv' ===")
```
