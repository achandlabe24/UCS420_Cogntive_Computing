##Q1

```python
# i. Extract individual digits, multiply by 10, and assign to list l
roll = "1247259315"
dig = [int(d) for d in roll]
multipd = [d * 10 for d in dig]

print(f"Digits: {dig}")
print(f"Multiplied: {multipd}")
l = multipd
print(f"Printing List l: {l}\n")
```

```python
# ii. Add two numbers using append() and insert() with your comment style
l.append(80)
print(f"After appending 80: {l}\n")
# 80 is added at the end
```

```python
l.insert(5, 50)
print(f"After inserting 50 at index 5: {l}\n")
# 50 is added at index 5
```

```python
# iii. Remove two elements using remove() and pop()
l.remove(50)
print(f"Removing 50: {l}\n")
# 50 is removed from the list
```

```python
l.pop()
print(f"Popping 1 time: {l}\n")
# The last element is removed from the list
```

```python
# iv. Sort l in ascending and descending order
l.sort()
print(f"Ascending Sort: {l}\n")

l.sort(reverse=True)
print(f"Descending Sort: {l}\n")
```

```python
# v. Slicing to print the first three and last three elements
print(f"First three elements: {l[:3]}")
print(f"Last three elements: {l[-3:]}\n")
```

```python
# vi. List comprehension to filter elements greater than the average
average_l = sum(l) / len(l)
elements_above_average = [x for x in l if x > average_l]
print(f"Elements greater than average ({average_l}): {elements_above_average}")
```

##Q2

```python
# Sample list L from Q1
L = [85, 92, 78, 92, 65, 89, 65, 94, 100]

# 2. Create a tuple named scores containing the first 8 marks
scores = tuple(L[:8])
print("Original scores tuple:", scores)
```

```python
# i. Find highest score/index, and lowest score/count
highest_score = max(scores)
highest_index = scores.index(highest_score)
lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print(f"Highest score: {highest_score} at index {highest_index}")
print(f"Lowest score: {lowest_score} appears {lowest_count} time(s)")
```

```python
# ii. Reverse the tuple and return the result as a list
# Tuples cannot be reversed in place because they are immutable; their elements and order cannot be altered after creation.
reversed_scores_list = list(reversed(scores))
print("Reversed as a list:", reversed_scores_list)
```

```python
# iii. Ask user for a score and check presence
try:
    user_score = int(input("Enter a score to search: "))
    if user_score in scores:
        print(f"Score {user_score} first appears at index {scores.index(user_score)}.")
    else:
        print(f"Score {user_score} is not present in the tuple.")
except ValueError:
    print("Invalid input. Please enter an integer.")
```

```python
# iv. Attempt to change one element directly and handle the error
try:
    scores[0] = 100
except TypeError as e:
    print(f"Caught expected error: {e}")
    # Tuples are immutable and throw a TypeError on assignment, whereas lists are mutable and allow direct element modification.
```

```python
# v. Unpack the tuple using the * operator
first, second, *remaining = scores
print(f"First: {first}, Second: {second}, Remaining: {remaining}")
```

```python
#vi. Extra
# Unpack directly into two variables by nesting the starred expression
(*first_three, rest) = (scores[:3], scores[3:])

print("First three:", first_three)
print("Rest of the list:", rest)
```
